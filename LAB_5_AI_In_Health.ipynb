```python
# ================================================================
# CSET343 - AI IN HEALTHCARE
# LAB ASSIGNMENT - 5
# MULTICLASS CLASSIFICATION
# DATASET: DERMATOLOGY
# ================================================================

# Install required library
!pip -q install ucimlrepo


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    roc_auc_score
)

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")


# ================================================================
# 2. DATA ACQUISITION
# ================================================================

print("=" * 70)
print("DATA ACQUISITION")
print("=" * 70)

# Load Dermatology dataset from UCI
dermatology = fetch_ucirepo(id=33)

X = dermatology.data.features
y = dermatology.data.targets

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFirst 5 rows of features:")
display(X.head())

print("\nFirst 5 rows of target:")
display(y.head())


# ================================================================
# 3. CREATE DATAFRAME
# ================================================================

df = pd.concat([X, y], axis=1)

print("\nDataset shape:", df.shape)

print("\nDataset:")
display(df.head())


# ================================================================
# 4. BASIC INFORMATION
# ================================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


# ================================================================
# 5. SUMMARY STATISTICS
# ================================================================

print("\n" + "=" * 70)
print("SUMMARY STATISTICS")
print("=" * 70)

summary = df.describe().T

summary["median"] = df.median(numeric_only=True)

summary = summary[
    [
        "count",
        "mean",
        "median",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]

display(summary)


# ================================================================
# 6. CHECK MISSING VALUES
# ================================================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

# Replace ? with NaN
df = df.replace("?", np.nan)

# Convert columns to numeric
for column in df.columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

missing = df.isnull().sum()

print(missing)

print("\nTotal missing values:", missing.sum())


# ================================================================
# 7. HANDLE MISSING VALUES
# ================================================================

print("\n" + "=" * 70)
print("HANDLING MISSING VALUES")
print("=" * 70)

# Fill missing values with median
for column in df.columns:

    if df[column].isnull().sum() > 0:

        df[column] = df[column].fillna(
            df[column].median()
        )

print("Missing values after handling:")
print(df.isnull().sum().sum())


# ================================================================
# 8. CHECK DUPLICATES
# ================================================================

print("\n" + "=" * 70)
print("DUPLICATE VALUES")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)

# Remove duplicates
if duplicates > 0:

    df = df.drop_duplicates()

    print(
        "Duplicates removed."
    )

else:

    print(
        "No duplicate rows found."
    )


# ================================================================
# 9. IDENTIFY FEATURES AND TARGET
# ================================================================

target_column = y.columns[0]

feature_columns = [
    column
    for column in df.columns
    if column != target_column
]

print("\nTarget column:", target_column)

print("Number of features:", len(feature_columns))


# ================================================================
# 10. OUTLIER DETECTION USING IQR
# ================================================================

print("\n" + "=" * 70)
print("OUTLIER DETECTION")
print("=" * 70)

outlier_results = []

for column in feature_columns:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append(
        {
            "Feature": column,
            "Outliers": len(outliers)
        }
    )

outlier_df = pd.DataFrame(outlier_results)

display(
    outlier_df.sort_values(
        "Outliers",
        ascending=False
    )
)


# ================================================================
# 11. BOX PLOT
# ================================================================

plt.figure(
    figsize=(20, 8)
)

df[feature_columns].boxplot()

plt.title(
    "Box Plot of Dermatology Features"
)

plt.xlabel("Features")

plt.ylabel("Values")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()


# ================================================================
# 12. HISTOGRAMS
# ================================================================

df[feature_columns].hist(
    figsize=(20, 20),
    bins=20
)

plt.suptitle(
    "Distribution of Dermatology Features",
    fontsize=18
)

plt.tight_layout()

plt.show()


# ================================================================
# 13. CORRELATION HEATMAP
# ================================================================

plt.figure(
    figsize=(20, 16)
)

correlation = df[
    feature_columns
].corr()

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap"
)

plt.tight_layout()

plt.show()


# ================================================================
# 14. CLASS DISTRIBUTION
# ================================================================

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)

class_counts = (
    df[target_column]
    .value_counts()
    .sort_index()
)

print(class_counts)


plt.figure(
    figsize=(8, 6)
)

sns.countplot(
    x=df[target_column],
    order=sorted(
        df[target_column].unique()
    )
)

plt.title(
    "Class Distribution"
)

plt.xlabel(
    "Disease Class"
)

plt.ylabel(
    "Number of Samples"
)

plt.show()


# ================================================================
# 15. CLASS DISTRIBUTION PERCENTAGE
# ================================================================

class_percentage = (
    df[target_column]
    .value_counts(
        normalize=True
    )
    .sort_index()
    * 100
)

print("\nClass distribution percentage:")

display(
    class_percentage
)


# ================================================================
# 16. SEPARATE FEATURES AND TARGET
# ================================================================

X = df[feature_columns]

y = df[target_column]


# ================================================================
# 17. ENCODE TARGET VARIABLE
# ================================================================

print("\n" + "=" * 70)
print("TARGET ENCODING")
print("=" * 70)

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print(
    "Original classes:"
)

print(
    label_encoder.classes_
)

print(
    "\nEncoded classes:"
)

print(
    np.unique(y_encoded)
)


# ================================================================
# 18. TRAIN TEST SPLIT
# ================================================================

print("\n" + "=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y_encoded,

    test_size=0.20,

    random_state=42,

    stratify=y_encoded
)

print(
    "Training samples:",
    X_train.shape[0]
)

print(
    "Testing samples:",
    X_test.shape[0]
)


# ================================================================
# 19. STANDARDIZATION
# ================================================================

print("\n" + "=" * 70)
print("FEATURE STANDARDIZATION")
print("=" * 70)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print(
    "Features standardized successfully."
)


# ================================================================
# 20. CREATE ML MODELS
# ================================================================

print("\n" + "=" * 70)
print("CREATING MACHINE LEARNING MODELS")
print("=" * 70)

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000,
            random_state=42
        ),

    "KNN":
        KNeighborsClassifier(
            n_neighbors=5
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ),

    "SVM":
        SVC(
            probability=True,
            random_state=42
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        )
}


# ================================================================
# 21. TRAIN ALL MODELS
# ================================================================

print("\n" + "=" * 70)
print("MODEL TRAINING")
print("=" * 70)

trained_models = {}

for name, model in models.items():

    print(
        "Training:",
        name
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    trained_models[name] = model

print(
    "\nAll models trained successfully."
)


# ================================================================
# 22. MODEL EVALUATION
# ================================================================

print("\n" + "=" * 70)
print("MODEL EVALUATION")
print("=" * 70)

results = []

for name, model in trained_models.items():

    y_pred = model.predict(
        X_test_scaled
    )

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        average="weighted"
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="weighted"
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="weighted"
    )

    results.append(
        {
            "Model": name,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1 Score": f1
        }
    )


results_df = pd.DataFrame(
    results
)


# ================================================================
# 23. CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORTS")
print("=" * 70)

for name, model in trained_models.items():

    y_pred = model.predict(
        X_test_scaled
    )

    print("\n")
    print("=" * 70)
    print(name)
    print("=" * 70)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=[
                str(c)
                for c in label_encoder.classes_
            ]
        )
    )


# ================================================================
# 24. CONFUSION MATRICES
# ================================================================

print("\n" + "=" * 70)
print("CONFUSION MATRICES")
print("=" * 70)

fig, axes = plt.subplots(
    2,
    3,
    figsize=(18, 12)
)

axes = axes.ravel()

for i, (name, model) in enumerate(
    trained_models.items()
):

    y_pred = model.predict(
        X_test_scaled
    )

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=axes[i]
    )

    axes[i].set_title(
        name
    )

    axes[i].set_xlabel(
        "Predicted Class"
    )

    axes[i].set_ylabel(
        "Actual Class"
    )

# Hide empty subplot
axes[-1].axis("off")

plt.tight_layout()

plt.show()


# ================================================================
# 25. MULTICLASS ROC-AUC
# ================================================================

print("\n" + "=" * 70)
print("ROC-AUC")
print("=" * 70)

n_classes = len(
    np.unique(y_encoded)
)

# Convert test labels into one-hot format
y_test_bin = label_binarize(
    y_test,
    classes=np.arange(n_classes)
)

auc_scores = {}

plt.figure(
    figsize=(12, 8)
)

for name, model in trained_models.items():

    # Probability prediction
    y_score = model.predict_proba(
        X_test_scaled
    )

    # Macro AUC
    macro_auc = roc_auc_score(
        y_test_bin,
        y_score,
        multi_class="ovr",
        average="macro"
    )

    auc_scores[name] = macro_auc

    # Micro-average ROC
    fpr_micro, tpr_micro, _ = roc_curve(
        y_test_bin.ravel(),
        y_score.ravel()
    )

    plt.plot(
        fpr_micro,
        tpr_micro,
        linewidth=2,
        label=(
            f"{name} "
            f"(AUC = {macro_auc:.3f})"
        )
    )


# Random classifier line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Multiclass ROC Curve - One vs Rest"
)

plt.legend()

plt.grid()

plt.show()


# ================================================================
# 26. AUC TABLE
# ================================================================

auc_df = pd.DataFrame(
    auc_scores.items(),
    columns=[
        "Model",
        "ROC-AUC"
    ]
)

print("\nROC-AUC Scores:")

display(
    auc_df
)


# ================================================================
# 27. FINAL RESULTS TABLE
# ================================================================

print("\n" + "=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)

final_results = results_df.merge(
    auc_df,
    on="Model"
)

display(
    final_results.round(4)
)


# ================================================================
# 28. PERFORMANCE COMPARISON GRAPH
# ================================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]

final_results.set_index(
    "Model"
)[metrics].plot(
    kind="bar",
    figsize=(15, 8)
)

plt.title(
    "Comparison of All Machine Learning Models"
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=30
)

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ================================================================
# 29. BEST MODEL FOR EACH METRIC
# ================================================================

print("\n" + "=" * 70)
print("BEST MODEL FOR EACH METRIC")
print("=" * 70)

for metric in metrics:

    index = final_results[
        metric
    ].idxmax()

    model_name = final_results.loc[
        index,
        "Model"
    ]

    score = final_results.loc[
        index,
        metric
    ]

    print(
        f"{metric}: "
        f"{model_name} "
        f"({score:.4f})"
    )


# ================================================================
# 30. CLASS IMBALANCE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("CLASS IMBALANCE ANALYSIS")
print("=" * 70)

max_class = class_counts.max()
min_class = class_counts.min()

imbalance_ratio = (
    max_class / min_class
)

print(
    "Largest class:",
    max_class
)

print(
    "Smallest class:",
    min_class
)

print(
    "Imbalance ratio:",
    round(
        imbalance_ratio,
        2
    )
)

if imbalance_ratio > 2:

    print(
        "\nThe dataset shows some class imbalance."
    )

    print(
        "This can affect model training because "
        "the model may perform better on majority "
        "classes than minority classes."
    )

else:

    print(
        "\nThe class distribution is relatively balanced."
    )


# ================================================================
# 31. FINAL INTERPRETATION
# ================================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

print("""
1. The Dermatology dataset was used for multiclass classification.

2. The dataset contains multiple classes representing different
   dermatological diseases.

3. Data exploration was performed using summary statistics,
   missing-value analysis, duplicate detection, outlier detection,
   histograms, box plots, correlation heatmap and class-distribution
   visualization.

4. Missing values, if present, were handled using the median of
   the corresponding feature.

5. Duplicate records were identified and removed if present.

6. The target variable was encoded using LabelEncoder.

7. The dataset was divided into training and testing sets using an
   80:20 stratified split.

8. StandardScaler was used to standardize the feature values.

9. Five machine learning algorithms were implemented:
   Logistic Regression, KNN, Random Forest, SVM and Decision Tree.

10. The models were evaluated using accuracy, precision, recall,
    F1-score, confusion matrix and ROC-AUC.

11. The confusion matrices show the correctly and incorrectly
    classified samples for each disease class.

12. ROC-AUC was calculated using the One-vs-Rest approach because
    this is a multiclass classification problem.

13. The final comparison table shows the performance of all
    implemented machine learning models.

14. The model performance should be interpreted using multiple
    metrics rather than accuracy alone, especially if the classes
    are imbalanced.
""")


# ================================================================
# 32. FINAL RESULT
# ================================================================

print("\n" + "=" * 70)
print("LAB ASSIGNMENT COMPLETED")
print("=" * 70)

print(
    "\nFinal Results:"
)

display(
    final_results.round(4)
)
```
