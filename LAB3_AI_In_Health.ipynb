```python
# ============================================================
# LAB ASSIGNMENT - 3
# AI IN HEALTHCARE (CSET343)
# Data Acquisition, Cleaning, Preprocessing & Analysis
# Tabular + Text + Image + Signal Data
# ============================================================

# =========================
# INSTALL REQUIRED LIBRARIES
# =========================

!pip install -q wfdb kaggle nltk pydicom

# =========================
# IMPORT LIBRARIES
# =========================

import os
import re
import string
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings('ignore')

print("All imports done!")

# ============================================================
# PART A - TABULAR DATA
# PIMA INDIANS DIABETES DATASET
# ============================================================

print("\n" + "="*70)
print("PART A - TABULAR DATA")
print("="*70)

# Dataset acquisition
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv'

cols = [
    'Pregnancies',
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI',
    'DiabetesPedigreeFunction',
    'Age',
    'Outcome'
]

df = pd.read_csv(url, names=cols)

print("Shape:", df.shape)

display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDescriptive Statistics:")
display(df.describe())

# ------------------------------------------------------------
# Replace impossible zero values with NaN
# ------------------------------------------------------------

zero_not_allowed = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]

for col in zero_not_allowed:
    df[col] = df[col].replace(0, np.nan)

print("\nMissing values after replacing zeros:")
print(df.isnull().sum())

# ------------------------------------------------------------
# Median imputation
# ------------------------------------------------------------

for col in zero_not_allowed:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

print("\nTotal missing values after imputation:")
print(df.isnull().sum().sum())

# ------------------------------------------------------------
# Outlier removal using IQR
# ------------------------------------------------------------

def remove_outliers_iqr(data, col):

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return data[
        (data[col] >= lower) &
        (data[col] <= upper)
    ]

df_clean = df.copy()

for col in ['Glucose', 'BMI', 'Insulin']:
    df_clean = remove_outliers_iqr(df_clean, col)

print("\nShape before outlier removal:", df.shape)
print("Shape after outlier removal:", df_clean.shape)

# ------------------------------------------------------------
# Feature distributions
# ------------------------------------------------------------

fig, axes = plt.subplots(3, 3, figsize=(14, 10))
axes = axes.flatten()

for i, col in enumerate(cols[:-1]):

    axes[i].hist(
        df_clean[df_clean['Outcome'] == 0][col],
        alpha=0.6,
        label='No Diabetes',
        color='steelblue',
        bins=20
    )

    axes[i].hist(
        df_clean[df_clean['Outcome'] == 1][col],
        alpha=0.6,
        label='Diabetes',
        color='salmon',
        bins=20
    )

    axes[i].set_title(col)
    axes[i].legend(fontsize=7)

axes[-1].axis('off')

plt.suptitle(
    'Feature Distributions by Diabetes Outcome',
    fontsize=13,
    y=1.01
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Correlation heatmap
# ------------------------------------------------------------

plt.figure(figsize=(9, 7))

sns.heatmap(
    df_clean.corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Statistical Tests
# ------------------------------------------------------------

from scipy import stats

# Independent Samples t-test

group0 = df_clean[
    df_clean['Outcome'] == 0
]['Glucose']

group1 = df_clean[
    df_clean['Outcome'] == 1
]['Glucose']

t_stat, p_val = stats.ttest_ind(
    group0,
    group1
)

print("\n=== Independent Samples t-test ===")
print("Glucose vs Diabetes Outcome")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.6f}")

if p_val < 0.05:
    print("Result: Statistically significant difference (reject H0)")
else:
    print("Result: No significant difference (fail to reject H0)")

# ------------------------------------------------------------
# Chi-Square Test
# ------------------------------------------------------------

df_clean['AgeGroup'] = pd.cut(
    df_clean['Age'],
    bins=[20, 35, 50, 100],
    labels=['Young', 'Middle', 'Senior']
)

contingency = pd.crosstab(
    df_clean['AgeGroup'],
    df_clean['Outcome']
)

chi2, p, dof, expected = stats.chi2_contingency(
    contingency
)

print("\n=== Chi-Square Test ===")
print("Age Group vs Diabetes Outcome")
print(f"Chi2 statistic: {chi2:.4f}")
print(f"P-value: {p:.6f}")
print(f"Degrees of freedom: {dof}")

if p < 0.05:
    print("Result: Age group and outcome are dependent (reject H0)")
else:
    print("Result: No significant association (fail to reject H0)")

# ------------------------------------------------------------
# One-Way ANOVA
# ------------------------------------------------------------

young_bmi = df_clean[
    df_clean['AgeGroup'] == 'Young'
]['BMI']

middle_bmi = df_clean[
    df_clean['AgeGroup'] == 'Middle'
]['BMI']

senior_bmi = df_clean[
    df_clean['AgeGroup'] == 'Senior'
]['BMI']

f_stat, p_val_anova = stats.f_oneway(
    young_bmi,
    middle_bmi,
    senior_bmi
)

print("\n=== One-Way ANOVA ===")
print("BMI across Age Groups")
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_val_anova:.6f}")

if p_val_anova < 0.05:
    print("Result: Significant difference in BMI across age groups (reject H0)")
else:
    print("Result: No significant difference (fail to reject H0)")

# ------------------------------------------------------------
# Feature Engineering
# ------------------------------------------------------------

from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

df_clean['Glucose_BMI'] = (
    df_clean['Glucose'] *
    df_clean['BMI']
)

df_clean['Insulin_per_Glucose'] = (
    df_clean['Insulin'] /
    (df_clean['Glucose'] + 1e-5)
)

df_clean['Age_Pregnancies'] = (
    df_clean['Age'] *
    df_clean['Pregnancies']
)

feature_cols = [
    c for c in df_clean.columns
    if c not in ['Outcome', 'AgeGroup']
]

X = df_clean[feature_cols]
y = df_clean['Outcome']

# Feature selection

selector = SelectKBest(
    score_func=f_classif,
    k=6
)

X_selected = selector.fit_transform(
    X,
    y
)

selected_features = X.columns[
    selector.get_support()
].tolist()

print("\nTop 6 selected features:")
print(selected_features)

# Feature scaling

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X_selected
)

print("\nPreprocessed tensor shape:")
print(X_scaled.shape)

print("\nMean approximately 0:")
print(X_scaled.mean(axis=0).round(3))

print("\nStd approximately 1:")
print(X_scaled.std(axis=0).round(3))


# ============================================================
# PART B - TEXTUAL DATA
# MTSAMPLES MEDICAL TRANSCRIPTIONS
# ============================================================

print("\n" + "="*70)
print("PART B - TEXTUAL DATA")
print("="*70)

# ------------------------------------------------------------
# Dataset acquisition
# ------------------------------------------------------------

# FIXED MTSAMPLES DATASET URL
df_text = pd.read_csv(
    'https://raw.githubusercontent.com/aneuraz/mtsamplesFR/master/data/mtsamples.csv'
)

print("Shape:", df_text.shape)

display(df_text.head(3))

print("\nColumns:")
print(df_text.columns.tolist())

print("\nMissing values:")
print(df_text.isnull().sum())

# ------------------------------------------------------------
# Select required columns
# ------------------------------------------------------------

df_text = df_text[
    [
        'medical_specialty',
        'transcription',
        'description'
    ]
].dropna(
    subset=['transcription']
)

df_text['medical_specialty'] = (
    df_text['medical_specialty']
    .astype(str)
    .str.strip()
)

print("\nCleaned shape:")
print(df_text.shape)

print("\nTop Medical Specialties:")

print(
    df_text['medical_specialty']
    .value_counts()
    .head(10)
)

# ------------------------------------------------------------
# PHI Removal
# ------------------------------------------------------------

def remove_phi(text):

    text = str(text)

    # Date formats
    text = re.sub(
        r'\b\d{1,2}/\d{1,2}/\d{2,4}\b',
        '[DATE]',
        text
    )

    text = re.sub(
        r'\b(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)'
        r'[a-z]*\.?\s+\d{1,2},?\s+\d{4}\b',
        '[DATE]',
        text,
        flags=re.IGNORECASE
    )

    # Age
    text = re.sub(
        r'\b\d{1,3}[\-\s]?year[\-\s]?old\b',
        '[AGE]-year-old',
        text,
        flags=re.IGNORECASE
    )

    # Phone number
    text = re.sub(
        r'\(\d{3}\)\s*\d{3}[\-\.]\d{4}',
        '[PHONE]',
        text
    )

    # Extra whitespace
    text = re.sub(
        r'\s+',
        ' ',
        text
    ).strip()

    return text


df_text['transcription_clean'] = (
    df_text['transcription']
    .apply(remove_phi)
)

print("\nSample after PHI removal:")

print(
    df_text['transcription_clean']
    .iloc[0][:400]
)

# ------------------------------------------------------------
# NLTK setup
# ------------------------------------------------------------

import nltk

nltk.download(
    'stopwords',
    quiet=True
)

nltk.download(
    'punkt',
    quiet=True
)

# IMPORTANT FIX FOR COLAB / NEW NLTK
nltk.download(
    'punkt_tab',
    quiet=True
)

nltk.download(
    'wordnet',
    quiet=True
)

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

stop_words = set(
    stopwords.words('english')
)

lemmatizer = WordNetLemmatizer()

# ------------------------------------------------------------
# Text preprocessing
# ------------------------------------------------------------

def preprocess_text(text):

    text = text.lower()

    text = text.translate(
        str.maketrans(
            '',
            '',
            string.punctuation
        )
    )

    tokens = word_tokenize(text)

    tokens = [
        lemmatizer.lemmatize(t)
        for t in tokens
        if t not in stop_words
        and len(t) > 2
    ]

    return ' '.join(tokens)


df_text['tokens'] = (
    df_text['transcription_clean']
    .apply(preprocess_text)
)

print("\nSample processed tokens:")

print(
    df_text['tokens']
    .iloc[0][:300]
)

# ------------------------------------------------------------
# TF-IDF
# ------------------------------------------------------------

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=500,
    ngram_range=(1, 2)
)

X_tfidf = tfidf.fit_transform(
    df_text['tokens']
)

print("\nTF-IDF matrix shape:")
print(X_tfidf.shape)

# ------------------------------------------------------------
# Top terms for selected specialties
# ------------------------------------------------------------

specialties = [
    'Surgery',
    'Cardiovascular / Pulmonary'
]

feature_names = (
    tfidf.get_feature_names_out()
)

for spec in specialties:

    subset = df_text[
        df_text['medical_specialty'] == spec
    ]['tokens']

    if len(subset) == 0:
        continue

    vec = TfidfVectorizer(
        max_features=10
    )

    vec.fit(subset)

    print(
        f"\nTop terms in [{spec}]:"
    )

    print(
        vec.get_feature_names_out()
        .tolist()
    )

# ------------------------------------------------------------
# Text length analysis
# ------------------------------------------------------------

df_text['text_length'] = (
    df_text['transcription_clean']
    .apply(
        lambda x: len(x.split())
    )
)

plt.figure(figsize=(10, 4))

plt.hist(
    df_text['text_length'],
    bins=50,
    color='mediumpurple',
    edgecolor='white'
)

plt.xlabel('Word Count')
plt.ylabel('Frequency')

plt.title(
    'Distribution of Transcription Length'
)

plt.axvline(
    df_text['text_length'].median(),
    color='red',
    linestyle='--',
    label='Median'
)

plt.legend()

plt.show()

print(
    df_text['text_length']
    .describe()
    .round(1)
)

# ------------------------------------------------------------
# Medical specialty distribution
# ------------------------------------------------------------

top_specs = (
    df_text['medical_specialty']
    .value_counts()
    .head(12)
)

plt.figure(figsize=(11, 5))

top_specs.plot(
    kind='bar',
    color='teal',
    edgecolor='white'
)

plt.title(
    'Top 12 Medical Specialties'
)

plt.ylabel('Count')

plt.xticks(
    rotation=45,
    ha='right'
)

plt.tight_layout()

plt.show()


# ============================================================
# PART C - IMAGE DATA
# CHEST X-RAY / PNEUMONIA
# ============================================================

print("\n" + "="*70)
print("PART C - IMAGE DATA")
print("="*70)

from PIL import Image
import torchvision.transforms as transforms
import torch

# ------------------------------------------------------------
# Dataset directory structure
# ------------------------------------------------------------

os.makedirs(
    'chest_xray/train/NORMAL',
    exist_ok=True
)

os.makedirs(
    'chest_xray/train/PNEUMONIA',
    exist_ok=True
)

TRAIN_DIR = 'chest_xray/train'

print("Directory structure ready.")

print(
    "\nNOTE:"
    "\nFor the actual Chest X-Ray dataset, download the"
    "\nChest X-Ray Images (Pneumonia) dataset and place"
    "\nthe images inside the NORMAL and PNEUMONIA folders."
)

# ------------------------------------------------------------
# Image preprocessing
# ------------------------------------------------------------

IMG_SIZE = 224

train_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],

        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])

val_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],

        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])

print(
    "Image transform pipelines defined."
)

# ------------------------------------------------------------
# ImageFolder setup
# ------------------------------------------------------------

from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader

# Uncomment these lines after putting the real images
# inside the folders.

# train_dataset = ImageFolder(
#     root=TRAIN_DIR,
#     transform=train_transform
# )

# train_loader = DataLoader(
#     train_dataset,
#     batch_size=32,
#     shuffle=True,
#     num_workers=2
# )

# ------------------------------------------------------------
# Demo image for preprocessing demonstration
# ------------------------------------------------------------

demo_img = Image.fromarray(
    np.random.randint(
        0,
        255,
        (300, 300),
        dtype=np.uint8
    )
)

demo_img_rgb = demo_img.convert(
    'RGB'
)

tensor = val_transform(
    demo_img_rgb
)

print(
    "\nInput image shape:",
    demo_img_rgb.size
)

print(
    "Output tensor shape:",
    tensor.shape
)

print(
    "Tensor min/max:",
    tensor.min().item(),
    tensor.max().item()
)

# ------------------------------------------------------------
# DICOM PHI anonymization
# ------------------------------------------------------------

import pydicom

PHI_TAGS = [

    'PatientName',
    'PatientID',
    'PatientBirthDate',
    'PatientSex',
    'PatientAge',
    'InstitutionName',
    'ReferringPhysicianName',
    'PatientAddress'
]

def anonymize_dicom(
    dcm_path,
    out_path
):

    ds = pydicom.dcmread(
        dcm_path
    )

    for tag in PHI_TAGS:

        if hasattr(ds, tag):

            delattr(
                ds,
                tag
            )

    ds.save_as(
        out_path
    )

    print(
        f'Anonymized: {dcm_path} -> {out_path}'
    )

print(
    "\nDICOM anonymization function ready."
)

# ------------------------------------------------------------
# Image augmentation demonstration
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    5,
    figsize=(16, 4)
)

aug_transforms = [

    (
        'Original',
        transforms.Compose([
            transforms.Resize(
                (224, 224)
            ),
            transforms.ToTensor()
        ])
    ),

    (
        'H-Flip',
        transforms.Compose([
            transforms.Resize(
                (224, 224)
            ),
            transforms.RandomHorizontalFlip(
                p=1.0
            ),
            transforms.ToTensor()
        ])
    ),

    (
        'Rotation',
        transforms.Compose([
            transforms.Resize(
                (224, 224)
            ),
            transforms.RandomRotation(
                15
            ),
            transforms.ToTensor()
        ])
    ),

    (
        'Brightness',
        transforms.Compose([
            transforms.Resize(
                (224, 224)
            ),
            transforms.ColorJitter(
                brightness=0.5
            ),
            transforms.ToTensor()
        ])
    ),

    (
        'Zoom',
        transforms.Compose([
            transforms.Resize(
                (256, 256)
            ),
            transforms.CenterCrop(
                224
            ),
            transforms.ToTensor()
        ])
    )
]

for ax, (name, tfm) in zip(
    axes,
    aug_transforms
):

    img_tensor = tfm(
        demo_img_rgb
    )

    ax.imshow(
        img_tensor.permute(
            1, 2, 0
        ).numpy()
    )

    ax.set_title(name)
    ax.axis('off')

plt.suptitle(
    'Augmentation Techniques Applied to Chest X-Ray',
    y=1.02
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# Chest X-Ray class distribution
# ------------------------------------------------------------

# Demonstration values corresponding to the dataset.
# Replace with actual counts when the real dataset is loaded.

class_counts = {
    'NORMAL': 1341,
    'PNEUMONIA': 3875
}

plt.figure(
    figsize=(5, 4)
)

plt.bar(
    class_counts.keys(),
    class_counts.values(),
    color=[
        'steelblue',
        'tomato'
    ]
)

plt.title(
    'Class Distribution - Chest X-Ray Dataset'
)

plt.ylabel(
    'Number of Images'
)

for k, v in class_counts.items():

    plt.text(
        list(class_counts.keys()).index(k),
        v + 50,
        str(v),
        ha='center',
        fontweight='bold'
    )

plt.tight_layout()
plt.show()

ratio = (
    class_counts['PNEUMONIA'] /
    class_counts['NORMAL']
)

print(
    f'\nClass imbalance ratio: {ratio:.2f}x more Pneumonia than Normal'
)

print(
    'Strategy: Use class weighting or oversampling during training'
)


# ============================================================
# PART D - SIGNAL DATA
# MIT-BIH ARRHYTHMIA DATABASE
# ============================================================

print("\n" + "="*70)
print("PART D - SIGNAL DATA")
print("="*70)

import wfdb

# ------------------------------------------------------------
# Download MIT-BIH record
# ------------------------------------------------------------

record_name = '100'

wfdb.dl_database(
    'mitdb',
    dl_dir='mit_bih',
    records=[record_name]
)

print(
    "MIT-BIH Record downloaded!"
)

# ------------------------------------------------------------
# Read ECG record
# ------------------------------------------------------------

record = wfdb.rdrecord(
    f'mit_bih/{record_name}'
)

annotation = wfdb.rdann(
    f'mit_bih/{record_name}',
    'atr'
)

ecg_signal = record.p_signal[:, 0]

fs = record.fs

print(
    f'Signal length: {len(ecg_signal)} samples'
)

print(
    f'Sampling frequency: {fs} Hz'
)

print(
    f'Duration: {len(ecg_signal)/fs:.1f} seconds'
)

print(
    f'Annotation types: {set(annotation.symbol)}'
)

# ------------------------------------------------------------
# Plot raw ECG
# ------------------------------------------------------------

duration_sec = 10

n_samples = int(
    duration_sec * fs
)

time_axis = (
    np.arange(n_samples) /
    fs
)

plt.figure(
    figsize=(14, 4)
)

plt.plot(
    time_axis,
    ecg_signal[:n_samples],
    color='navy',
    linewidth=0.8
)

plt.xlabel(
    'Time (seconds)'
)

plt.ylabel(
    'Amplitude (mV)'
)

plt.title(
    f'MIT-BIH Record {record_name} - '
    f'Lead MLII (First {duration_sec}s)'
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# ECG filtering
# ------------------------------------------------------------

from scipy.signal import (
    butter,
    filtfilt,
    iirnotch
)

def bandpass_filter(
    signal,
    lowcut=0.5,
    highcut=40.0,
    fs=360,
    order=4
):

    nyq = fs / 2

    low = lowcut / nyq
    high = highcut / nyq

    b, a = butter(
        order,
        [low, high],
        btype='band'
    )

    return filtfilt(
        b,
        a,
        signal
    )


def notch_filter(
    signal,
    freq=60.0,
    fs=360,
    Q=30
):

    b, a = iirnotch(
        freq / (fs / 2),
        Q
    )

    return filtfilt(
        b,
        a,
        signal
    )


ecg_bp = bandpass_filter(
    ecg_signal,
    fs=fs
)

ecg_clean = notch_filter(
    ecg_bp,
    fs=fs
)

print(
    "\nSignal filtered."
)

print(
    f'Raw standard deviation: '
    f'{ecg_signal.std():.4f}'
)

print(
    f'Clean standard deviation: '
    f'{ecg_clean.std():.4f}'
)

# ------------------------------------------------------------
# Raw vs filtered ECG
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2,
    1,
    figsize=(14, 6),
    sharex=True
)

n = int(
    5 * fs
)

t = (
    np.arange(n) /
    fs
)

axes[0].plot(
    t,
    ecg_signal[:n],
    color='gray',
    linewidth=0.7,
    label='Raw'
)

axes[0].set_title(
    'Raw ECG Signal'
)

axes[0].set_ylabel(
    'Amplitude (mV)'
)

axes[0].grid(
    alpha=0.3
)

axes[1].plot(
    t,
    ecg_clean[:n],
    color='steelblue',
    linewidth=0.7,
    label='Filtered'
)

axes[1].set_title(
    'Filtered ECG (Bandpass + Notch)'
)

axes[1].set_xlabel(
    'Time (s)'
)

axes[1].set_ylabel(
    'Amplitude (mV)'
)

axes[1].grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# ECG beat segmentation
# ------------------------------------------------------------

WINDOW_BEFORE = 90
WINDOW_AFTER = 110

WINDOW_SIZE = (
    WINDOW_BEFORE +
    WINDOW_AFTER
)

r_peaks = annotation.sample

labels = annotation.symbol

windows = []

beat_labels = []

for peak, label in zip(
    r_peaks,
    labels
):

    start = (
        peak -
        WINDOW_BEFORE
    )

    end = (
        peak +
        WINDOW_AFTER
    )

    if (
        start >= 0 and
        end <= len(ecg_clean)
    ):

        windows.append(
            ecg_clean[start:end]
        )

        beat_labels.append(
            label
        )

windows = np.array(
    windows
)

print(
    f'\nTotal beat windows: '
    f'{len(windows)}'
)

print(
    f'Window shape: '
    f'{windows.shape}'
)

# ------------------------------------------------------------
# Beat class distribution
# ------------------------------------------------------------

from collections import Counter

beat_counts = Counter(
    beat_labels
)

beat_df = pd.DataFrame(
    beat_counts.items(),
    columns=[
        'Beat Type',
        'Count'
    ]
).sort_values(
    'Count',
    ascending=False
)

print(
    "\nBeat Type Distribution:"
)

print(
    beat_df.to_string(
        index=False
    )
)

plt.figure(
    figsize=(8, 4)
)

plt.bar(
    beat_df['Beat Type'],
    beat_df['Count'],
    color='darkorange',
    edgecolor='white'
)

plt.title(
    'Beat Type Distribution - MIT-BIH Record 100'
)

plt.xlabel(
    'Beat Symbol'
)

plt.ylabel(
    'Count'
)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# Normalize ECG windows
# ------------------------------------------------------------

def normalize_windows(windows):

    mins = windows.min(
        axis=1,
        keepdims=True
    )

    maxs = windows.max(
        axis=1,
        keepdims=True
    )

    return (
        (windows - mins) /
        (
            maxs -
            mins +
            1e-8
        )
    )


windows_norm = normalize_windows(
    windows
)

print(
    "\nNormalized window statistics:"
)

print(
    f'Min: {windows_norm.min():.4f}'
)

print(
    f'Max: {windows_norm.max():.4f}'
)

# ------------------------------------------------------------
# Plot normal ECG beats
# ------------------------------------------------------------

normal_idx = [
    i
    for i, l in enumerate(beat_labels)
    if l == 'N'
][:5]

fig, axes = plt.subplots(
    1,
    5,
    figsize=(15, 3),
    sharey=True
)

for ax, idx in zip(
    axes,
    normal_idx
):

    ax.plot(
        windows_norm[idx],
        color='steelblue',
        linewidth=1
    )

    ax.set_title(
        f'Normal Beat #{idx}'
    )

    ax.set_xlabel(
        'Sample'
    )

    ax.set_ylim(
        0,
        1
    )

axes[0].set_ylabel(
    'Normalized Amplitude'
)

plt.suptitle(
    'Sample Normal (N) ECG Beats After Preprocessing',
    y=1.02
)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# ECG feature extraction
# ------------------------------------------------------------

def extract_ecg_features(
    window
):

    features = {

        'mean':
        window.mean(),

        'std':
        window.std(),

        'min':
        window.min(),

        'max':
        window.max(),

        'rms':
        np.sqrt(
            np.mean(
                window ** 2
            )
        ),

        'peak_to_peak':
        window.max() -
        window.min(),

        'skewness':
        pd.Series(
            window
        ).skew(),

        'kurtosis':
        pd.Series(
            window
        ).kurtosis(),

        'energy':
        np.sum(
            window ** 2
        ),

        'zero_crossings':
        (
            (
                window[:-1] *
                window[1:]
            ) < 0
        ).sum()
    }

    return features


feature_list = [
    extract_ecg_features(w)
    for w in windows_norm
]

df_ecg_feat = pd.DataFrame(
    feature_list
)

df_ecg_feat['label'] = (
    beat_labels
)

print(
    "\nECG feature matrix shape:"
)

print(
    df_ecg_feat.shape
)

display(
    df_ecg_feat.head()
)

# ------------------------------------------------------------
# FFT - Frequency domain analysis
# ------------------------------------------------------------

sample_beat = (
    windows_norm[
        normal_idx[0]
    ]
)

fft_vals = np.fft.rfft(
    sample_beat
)

freqs = np.fft.rfftfreq(
    WINDOW_SIZE,
    d=1/fs
)

power = (
    np.abs(
        fft_vals
    ) ** 2
)

plt.figure(
    figsize=(10, 4)
)

plt.subplot(
    1,
    2,
    1
)

plt.plot(
    sample_beat,
    color='steelblue'
)

plt.title(
    'Time Domain - Single Beat'
)

plt.xlabel(
    'Sample'
)

plt.ylabel(
    'Amplitude'
)

plt.subplot(
    1,
    2,
    2
)

plt.plot(
    freqs,
    power,
    color='darkorange'
)

plt.title(
    'Frequency Domain (FFT Power Spectrum)'
)

plt.xlabel(
    'Frequency (Hz)'
)

plt.ylabel(
    'Power'
)

plt.xlim(
    0,
    80
)

plt.tight_layout()

plt.show()


# ============================================================
# LAB COMPLETION
# ============================================================

print("\n" + "="*70)
print("LAB 3 COMPLETED SUCCESSFULLY")
print("="*70)

print("\nDatasets processed:")
print("1. Pima Indians Diabetes - Tabular")
print("2. MTSamples Medical Transcriptions - Text")
print("3. Chest X-Ray / Pneumonia - Image")
print("4. MIT-BIH Arrhythmia - ECG Signal")

print("\nAll major preprocessing, EDA, statistical analysis,")
print("feature engineering, feature selection, text processing,")
print("image augmentation, PHI handling and ECG signal")
print("processing steps have been executed.")
```
