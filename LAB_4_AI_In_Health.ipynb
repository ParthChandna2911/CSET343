```python
# ================================================================
# CSET343 - AI IN HEALTHCARE
# LAB ASSIGNMENT - 4
# LOGISTIC REGRESSION FOR BINARY CLASSIFICATION
# DATASET: BREAST CANCER WISCONSIN (DIAGNOSTIC)
# GOOGLE COLAB - COMPLETE CODE
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")


# ================================================================
# 2. LOAD DATASET
# ================================================================

print("=" * 70)
print("DATA ACQUISITION")
print("=" * 70)

# Load Breast Cancer Wisconsin Diagnostic dataset
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="Outcome"
)

print("Dataset loaded successfully!")

print("\nNumber of samples:", X.shape[0])
print("Number of features:", X.shape[1])

print("\nTarget classes:")
print(data.target_names)

print("\nFirst 5 rows:")
display(X.head())


# ================================================================
# 3. CREATE COMPLETE DATAFRAME
# ================================================================

df = X.copy()

df["Outcome"] = y

print("\nComplete dataset:")
display(df.head())


# ================================================================
# 4. DATASET INFORMATION
# ================================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


# ================================================================
# 5. SUMMARY STATISTICS
# Mean, Median, Standard Deviation, etc.
# ================================================================

print("\n" + "=" * 70)
print("SUMMARY STATISTICS")
print("=" * 70)

summary = df.describe().T

summary["median"] = df.median()

summary = summary[
    [
        "count",
        "mean",
        "median",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]

display(summary)


# ================================================================
# 6. CHECK MISSING VALUES
# ================================================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing_values = df.isnull().sum()

print(missing_values)

print(
    "\nTotal missing values:",
    missing_values.sum()
)


# ================================================================
# 7. HANDLE MISSING VALUES
# ================================================================

print("\n" + "=" * 70)
print("HANDLING MISSING VALUES")
print("=" * 70)

if missing_values.sum() == 0:

    print("No missing values found.")

else:

    # Fill numerical missing values with median
    for column in X.columns:

        if df[column].isnull().sum() > 0:

            df[column] = df[column].fillna(
                df[column].median()
            )

    print("Missing values handled using median.")


# ================================================================
# 8. CHECK DUPLICATES
# ================================================================

print("\n" + "=" * 70)
print("DUPLICATE VALUES")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(
    "Number of duplicate rows:",
    duplicate_count
)

if duplicate_count > 0:

    df = df.drop_duplicates()

    print("Duplicate rows removed.")

else:

    print("No duplicate rows found.")


# ================================================================
# 9. OUTLIER DETECTION USING IQR
# ================================================================

print("\n" + "=" * 70)
print("OUTLIER DETECTION")
print("=" * 70)

feature_columns = list(X.columns)

outlier_results = []

for column in feature_columns:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_results.append({
        "Feature": column,
        "Number of Outliers": len(outliers)
    })


outlier_df = pd.DataFrame(
    outlier_results
)

display(
    outlier_df.sort_values(
        "Number of Outliers",
        ascending=False
    )
)


# ================================================================
# 10. BOXPLOT
# ================================================================

print("\n" + "=" * 70)
print("BOX PLOT")
print("=" * 70)

plt.figure(
    figsize=(20, 8)
)

df[feature_columns].boxplot()

plt.title(
    "Box Plot of Breast Cancer Features"
)

plt.xlabel("Features")

plt.ylabel("Values")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()


# ================================================================
# 11. HISTOGRAMS
# ================================================================

print("\n" + "=" * 70)
print("HISTOGRAMS")
print("=" * 70)

df[feature_columns].hist(
    figsize=(20, 20),
    bins=20
)

plt.suptitle(
    "Distribution of Breast Cancer Features",
    fontsize=18
)

plt.tight_layout()

plt.show()


# ================================================================
# 12. CORRELATION MATRIX / HEATMAP
# ================================================================

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

correlation_matrix = df[
    feature_columns
].corr()

plt.figure(
    figsize=(18, 15)
)

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap"
)

plt.tight_layout()

plt.show()


# ================================================================
# 13. CLASS DISTRIBUTION
# ================================================================

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)

class_counts = df["Outcome"].value_counts()

print(class_counts)

print("\nClass names:")

print(
    "0 = Malignant"
)

print(
    "1 = Benign"
)


plt.figure(
    figsize=(7, 5)
)

sns.countplot(
    x=df["Outcome"]
)

plt.title(
    "Class Distribution"
)

plt.xlabel(
    "Outcome (0 = Malignant, 1 = Benign)"
)

plt.ylabel(
    "Number of Samples"
)

plt.show()


# ================================================================
# 14. CLASS DISTRIBUTION PERCENTAGE
# ================================================================

class_percentage = (
    df["Outcome"]
    .value_counts(normalize=True)
    * 100
)

print("\nClass distribution percentage:")

display(class_percentage)


# ================================================================
# 15. PREPARE X AND Y
# ================================================================

print("\n" + "=" * 70)
print("DATA PREPROCESSING")
print("=" * 70)

X = df.drop(
    columns=["Outcome"]
)

y = df["Outcome"]

print(
    "X shape:",
    X.shape
)

print(
    "y shape:",
    y.shape
)


# ================================================================
# 16. TRAIN-TEST SPLIT
# 80% TRAINING / 20% TESTING
# STRATIFIED SAMPLING
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)

print(
    "\nTraining samples:",
    X_train.shape[0]
)

print(
    "Testing samples:",
    X_test.shape[0]
)

print("\nTraining class distribution:")

print(
    y_train.value_counts()
)

print("\nTesting class distribution:")

print(
    y_test.value_counts()
)


# ================================================================
# 17. STANDARDIZATION
# ================================================================

print("\n" + "=" * 70)
print("FEATURE STANDARDIZATION")
print("=" * 70)

scaler = StandardScaler()

# IMPORTANT:
# Fit only on training data
X_train_scaled = scaler.fit_transform(
    X_train
)

# Only transform test data
X_test_scaled = scaler.transform(
    X_test
)

print(
    "Features standardized successfully."
)

print(
    "\nTraining mean after scaling:",
    np.round(
        X_train_scaled.mean(axis=0)[:5],
        4
    )
)

print(
    "Training standard deviation after scaling:",
    np.round(
        X_train_scaled.std(axis=0)[:5],
        4
    )
)


# ================================================================
# 18. CREATE LOGISTIC REGRESSION MODEL
# ================================================================

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION")
print("=" * 70)

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)


# ================================================================
# 19. TRAIN MODEL
# ================================================================

model.fit(
    X_train_scaled,
    y_train
)

print(
    "Logistic Regression model trained successfully."
)


# ================================================================
# 20. PREDICTION
# ================================================================

y_pred = model.predict(
    X_test_scaled
)

# Probability of positive class
y_prob = model.predict_proba(
    X_test_scaled
)[:, 1]


print(
    "\nFirst 10 predicted classes:"
)

print(
    y_pred[:10]
)

print(
    "\nFirst 10 predicted probabilities:"
)

print(
    np.round(
        y_prob[:10],
        4
    )
)


# ================================================================
# 21. ACCURACY
# ================================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)


# ================================================================
# 22. PRECISION
# ================================================================

precision = precision_score(
    y_test,
    y_pred
)


# ================================================================
# 23. RECALL
# ================================================================

recall = recall_score(
    y_test,
    y_pred
)


# ================================================================
# 24. F1 SCORE
# ================================================================

f1 = f1_score(
    y_test,
    y_pred
)


# ================================================================
# 25. ROC-AUC
# ================================================================

auc_score = roc_auc_score(
    y_test,
    y_prob
)


# ================================================================
# 26. DISPLAY METRICS
# ================================================================

print("\n" + "=" * 70)
print("MODEL EVALUATION")
print("=" * 70)

print(
    f"Accuracy  : {accuracy:.4f}"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall    : {recall:.4f}"
)

print(
    f"F1 Score  : {f1:.4f}"
)

print(
    f"ROC-AUC   : {auc_score:.4f}"
)


# ================================================================
# 27. CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Malignant",
            "Benign"
        ]
    )
)


# ================================================================
# 28. CONFUSION MATRIX
# ================================================================

print("\n" + "=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)


plt.figure(
    figsize=(7, 6)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Malignant",
        "Benign"
    ],
    yticklabels=[
        "Malignant",
        "Benign"
    ]
)

plt.title(
    "Confusion Matrix - Logistic Regression"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.show()


# ================================================================
# 29. EXTRACT TP, TN, FP, FN
# ================================================================

TN, FP, FN, TP = cm.ravel()

print("\nConfusion Matrix Components:")

print(
    "True Negative (TN):",
    TN
)

print(
    "False Positive (FP):",
    FP
)

print(
    "False Negative (FN):",
    FN
)

print(
    "True Positive (TP):",
    TP
)


# ================================================================
# 30. ROC CURVE
# ================================================================

print("\n" + "=" * 70)
print("ROC CURVE")
print("=" * 70)

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)


plt.figure(
    figsize=(8, 6)
)

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"Logistic Regression (AUC = {auc_score:.3f})"
)

# Random classifier
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve - Logistic Regression"
)

plt.legend()

plt.grid()

plt.show()


# ================================================================
# 31. MODEL COEFFICIENTS
# MODEL INTERPRETATION
# ================================================================

print("\n" + "=" * 70)
print("MODEL COEFFICIENTS")
print("=" * 70)

coefficients = pd.DataFrame({

    "Feature": X.columns,

    "Coefficient": model.coef_[0]

})

coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"]
    .abs()
)

coefficients = coefficients.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

display(
    coefficients.head(10)
)


# ================================================================
# 32. FINAL RESULTS TABLE
# ================================================================

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

results = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        auc_score
    ]
})

results["Score"] = results["Score"].round(4)

display(results)


# ================================================================
# 33. INTERPRETATION
# ================================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

print("""
1. Logistic Regression was used because the task is a binary
   classification problem.

2. The dataset was explored using summary statistics, missing-value
   analysis, duplicate detection, outlier detection, histograms,
   box plots, correlation heatmap and class distribution.

3. The target variable contains two classes:
   Malignant and Benign.

4. The dataset was divided into 80% training data and 20% testing
   data using stratified sampling.

5. StandardScaler was used to standardize the features. The scaler
   was fitted only on the training data and then applied to the
   testing data to avoid data leakage.

6. Logistic Regression was trained on the standardized training data.

7. Accuracy measures the overall percentage of correct predictions.

8. Precision measures how many of the samples predicted as positive
   were actually positive.

9. Recall measures how many of the actual positive samples were
   correctly identified.

10. F1-score provides a balance between precision and recall.

11. The confusion matrix shows True Positive, True Negative,
    False Positive and False Negative predictions.

12. The ROC curve shows the relationship between True Positive Rate
    and False Positive Rate at different classification thresholds.

13. ROC-AUC summarizes the model's ability to distinguish between
    the two classes.

14. In a medical diagnosis application, false negatives can be
    particularly important because a malignant case incorrectly
    classified as benign may be missed.
""")


# ================================================================
# 34. LAB COMPLETED
# ================================================================

print("\n" + "=" * 70)
print("LAB ASSIGNMENT 4 COMPLETED SUCCESSFULLY")
print("=" * 70)
```
